```

graph = {
    'S': ['A', 'B'],
    'A': [],
    'B': ['C', 'D'],
    'C': ['E', 'F'],
    'D': ['G' , 'H'],
    'H': ['E'],
    'F': [],
    'G': []
}

goal = 'E'



def DLS(node, depth, path):

    if node == goal:
        return path

    if depth == 0:
        return None

    for child in graph[node]:
        result = DLS(child, depth - 1, path + [child])

        if result is not None:
            return result

    return None



def IDS(start):

    depth = 0

    while True:
        print("Searching at depth:", depth)

        result = DLS(start, depth, [start])

        if result is not None:
            return result

        depth += 1



path = IDS('S')


print("M Afnaan Ali")
print("1wn24cs148")

print("\nGoal found!")
print("Path:", " -> ".join(path))
```
