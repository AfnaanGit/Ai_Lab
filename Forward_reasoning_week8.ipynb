```python

facts = {
    ("Barks", "Tommy"),
    ("HasFourLegs", "Tommy"),
    ("HasSharpTeeth", "Tommy")
}


def forward_chaining(facts):

    while True:
        new_facts = set()

        for f1 in facts:
            for f2 in facts:
                if (f1[0] == "Barks" and
                    f2[0] == "HasFourLegs" and
                    f1[1] == f2[1]):

                    new_facts.add(("Dog", f1[1]))


        for f in facts:
            if f[0] == "Dog":
                new_facts.add(("Mammal", f[1]))


        for f1 in facts:
            for f2 in facts:
                if (f1[0] == "Mammal" and
                    f2[0] == "HasSharpTeeth" and
                    f1[1] == f2[1]):

                    new_facts.add(("Carnivore", f1[1]))

        for f in facts:
            if f[0] == "Bird":
                new_facts.add(("CanFly", f[1]))


        new_facts = new_facts - facts

        if not new_facts:
            break

        facts.update(new_facts)

        print("New facts:", new_facts)

    print("\nFinal Knowledge Base:")
    for f in facts:
        print(f)


forward_chaining(facts)
```
