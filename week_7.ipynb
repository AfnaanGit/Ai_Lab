```python

def unify(x, y, subst):

    if x == y:
        return subst


    if isinstance(x, str) and x.islower():
        subst[x] = y
        return subst


    if isinstance(y, str) and y.islower():
        subst[y] = x
        return subst


    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return "FAILURE"

        if len(x) != len(y):
            return "FAILURE"


        for i in range(1, len(x)):
            result = unify(x[i], y[i], subst)

            if result == "FAILURE":
                return "FAILURE"

        return subst

    return "FAILURE"


# Main program
exp1 = ("Eats", "x", "Apple")
exp2 = ("Eats", "Riya", "y")

SUBST = {}

result = unify(exp1, exp2, SUBST)

if result == "FAILURE":
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)




```
