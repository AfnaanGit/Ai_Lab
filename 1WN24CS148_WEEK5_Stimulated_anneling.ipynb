```python

import random
import math


def cost(state):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    current = [random.randint(0, 7) for _ in range(8)]
    temperature = 100.0
    cooling = 0.99

    while temperature > 0.1:
        current_cost = cost(current)

        if current_cost == 0:
            print("Solution:", current)
            print("Cost:", current_cost)
            return

        neighbour = current[:]
        col = random.randint(0, 7)
        new_row = random.randint(0, 7)

        while new_row == neighbour[col]:
            new_row = random.randint(0, 7)

        neighbour[col] = new_row
        neighbour_cost = cost(neighbour)

        deltaE = current_cost - neighbour_cost

        if deltaE > 0:
            current = neighbour
        else:
            probability = math.exp(deltaE / temperature)

            if random.random() < probability:
                current = neighbour

        temperature = temperature * cooling

        print(current)

    print("Final state:", current)
    print("Cost:", cost(current))


    if cost(current) == 0:
        print("Solution found!")
    else:
        print("No solution found in this run. Try again.")


simulated_annealing()



```
